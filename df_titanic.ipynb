```python
df.rename(columns={
    'PassengerId': 'ID_Passageniro', 
    'Survived': 'Sobreviveu', 
    'Pclass': 'Classe', 
    'Name': 'Nome', 
    'Sex': 'Sexo', 
    'Age': 'Idade', 
    'SibSp': 'Membros_Familia',
    'Parch': 'Pais_Filhos_a_Bordo',
    'Ticket': 'Bilhete', 
    'Fare': 'Valor_Passagem', 
    'Cabin':'Cabine', 
    'Embarked': 'Portao_Embarque'
}, inplace=True)
df.head(3)
```

```python
dados = df.copy()
dados.head(3)
```

```python
dados.drop(['ID_Passageniro'], axis=1, inplace=True)
dados.head(3)
```

```python
dados[['Sobreviveu', 'Classe']]
```

###**Loc**

```python
#Número das linhas (Valor incial até o valor final) e número das colunas (passar uma lista)
dados.loc[15:35, ['Sobreviveu', 'Nome', 'Idade']]
```

```python
dados.loc[:500, ['Sobreviveu', 'Nome', 'Idade']]
```

###**Seleção por condição**

```python
dados.loc[dados['Classe'] == 1]
```

```python
dados.loc[dados['Sexo'] == 'female']
```

```python
dados.loc[dados['Idade'] < 18]
```

```python
dados.loc[(dados['Idade'] <= 18) & (dados['Sexo'] == 'male') & (dados['Sobreviveu'] != 0)]
```

```python
dados.loc[(dados['Idade'] <= 18) & (dados['Sexo'] == 'male')][['Nome', 'Portao_Embarque']]
```

```python
dados.loc[(dados['Idade'] <= 18) & (dados['Sexo'] == 'male') | (dados['Sexo'] == 'female')][['Nome', 'Portao_Embarque']]
```
