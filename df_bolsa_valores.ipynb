```python
import pandas as pd
import numpy as np
import yfinance as y
import matplotlib.pyplot as plt
```

```python
df = y.download('petr4.sa', period='3y')
df.sample(5)
```

```python
df.shape
```

```python
df.info()
```

###**Qual o valor de fechamento mais baixo do preços?**

```python
df['Close'].min().round(2)
```

###**Qual o valor de fechamento mais alto do preço?**

```python
df['Close'].max()
```

###**Qual o preço médio de fechamento?**

```python
df['Close'].mean().round(2)
```

###**Qual a maior queda do papel dentro do período?**

```python
df['Variacao'] = df['Close'].pct_change() * 100
df.head(3)
```

```python
df['Variacao'].min()
```

###**Qual o dia da pior queda do ativo?**

```python
filtro = df['Variacao'].min()
```

```python
df[df['Variacao'] == filtro]
```

###**Qual dia da maior alta**

```python
filtro_1 = df['Variacao'].max()
```

```python
df[df['Variacao'] == filtro_1]
```
