```python
excluir = ['CNPJ da Revenda', 'Nome da Rua', 'Complemento', 'Bairro', 'Cep', 'Valor de Compra']
df = df.drop(excluir, axis=1)
```

```python
df.head(3)
```

```python
df.isnull().sum()
```

```python
df['Valor de Venda'].max()
```

```python
def remove_virgula(x):
    try:
        x = str(x).replace(',', '.')
        return float(x)
    except:
        return np.NaN
```

```python
df['Valor de Venda'] = df['Valor de Venda'].map(remove_virgula)
df.head(3)
```

```python
df.info()
```

```python
df = df.drop(['Numero Rua'], axis=1)
```
